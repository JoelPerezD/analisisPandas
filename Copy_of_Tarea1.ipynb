Generar un dataset con la siguiente información en un conjunto de **n** registros

* Estatura de una persona

```python
#TODO Incluir las librerías necesarias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

##Generar dataset vacío

```python
#TODO Definir un dataset vacío
dataset = pd.DataFrame()
```

### Primeras dos columnas (llenado aleatorio)

* para el peso considerar un valor entero en el rango [55,120] kgs,
* para la estatura generar un valor flotante en el rango [1.5,2.0] mts.


```python
# Determinar cuántos registros se van a generar
registros = 20
```

```python
# prompt:  Generar la columna Peso con valores aleatorios en el rango [45,100]

dataset['Peso'] = np.random.randint(45, 101, size=registros)
dataset
```

```python
#TODO Generar la columna Estatura con valores aleatorios en el rango [1.5,2.0]
dataset['Estatura'] = np.random.uniform(1.5,2.0, size=registros)
dataset
```

###Columna IMC

El IMC se calcula como sigue

$IMC = \frac{peso}{estatura^2}$


```python
#TODO Calcular el IMC para cada registro
dataset['IMC']=dataset['Peso']/dataset['Estatura']**2
dataset
```

### Columna Clase

Considerar las siguientes clases ['Bajo peso', 'Peso normal', 'Sobrepeso', 'Obesidad']. Se asigna una clase a cada registro a partir del IMC como sigue

* $IMC<18$, se considera bajo peso
* $18<=IMC<24$ se considera peso normal
* $24<=IMC<29$ se considera sobrepeso
* $IMC>=29$, se considera obesidad

```python
#Definir las clases a utilizar
Clasificacion = ['Bajo peso', 'Normal', 'Sobrepeso','Obesidad']
dataset['Clase'] =pd.cut(dataset['IMC'], bins=[0, 18, 24, 29, np.inf], labels=Clasificacion)
dataset
```

##Gráfica pie

Mostra una gráfica de pie donde se muestra la distribución de las muestras por clase.

```python
#TODO Gráfica pie, con clases
dataset['Clase'].value_counts().plot(kind='pie')
plt.show()
```
